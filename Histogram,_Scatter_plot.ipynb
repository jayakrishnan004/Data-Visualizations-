# **HISTOGRAM** - important for viva
A graphical tool in data science used to show the frequency distribution of a single numerical, continuos attribute. It groups data values into adjacent intervals called **BINS** on the horizontal axis and plots the count or density of observations in each bin under vertical axis.

**NOTE**

When analyzing a histogram in EDA(exploratory data analysis) look at the following properties.

Shape/Distribution type :
1. Bell/Normal - data clusters symmetrically around a central peak, the mean roughly equals the median.
2. Right-Skewed(positive skewed) - a long tail extends to the right, mean is greater than the median.
3. Left-Skewed(negative skewed) - a long tail extends to the left, median is greater than the mean.
4. Bi-Model(multi model) - Two or more distinct peaks appear, suggesting the data contains two different underlined groups or processes.
5. Bars are roughly the same height, meaning values occur with similar frequency across the range.

Centre - where the bulk of the data lies, it gives you a sence of typical value or average of that attribute.

Spread - How wide the distribution is.
* Narrow histogram - Data points are tightly packed(Low variance).
* Wide histogram - Values are spread out(High variance).


**OUTLIERS AND GAPS**

Outliers - Isolated bars sitting far away from the main cluster of data.
Gaps - empty bins inside the range where no observations fall, indicated missing data segments.

Impact of BIN WIDTH :
Too wide bins hide local patterns and smooth out important details.
Too narrow bins create a jagged(uneven shape) look dominated by random noise.

```python
import pandas as pd
file = "/content/drive/MyDrive/DSA - ICTAK/DATASETS/auto-mpg.csv"
df = pd.read_csv(file)
df.head()
```

```python
print(df.columns)
```

```python
import matplotlib.pyplot as plt

plt.hist(df['mpg'])
plt.title("Histogram of MPG")

```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(8, 8))
plt.hist(df["mpg"],color='red',rwidth=0.5, cumulative = True)
# The relative width of the bar as a fraction of the bin width
# Histogram is computed where each bins gives the count in that bin + all bins for smaller values.
# The last bin gives the total number of data points.
# That means cumulative = True changes the histogram from a standard frequency count to a cumulative count.
plt.xlabel("mpg")
plt.ylabel("frequency")
plt.title("Histogram of MPG", fontsize=16)
plt.show()

```

# **SCATTER PLOT**
A data display that uses cartesian coordinates to plot and visualize the relationship between two numerical variables.
* X-axis represents the independent variable and Y-axis represents the dependent variable.
* As the value on X-axis increase value on Y-axis also increase - positive coordination.
* As the value on X-axis increase value on Y-axis decreases - negative coordination.
* Points are randomly scattered with no pattern.

USES in Data Science :
* correlation analysis
* outlier detection - spotting the data points that fall far outside the normal cluster.
* Cluster identification - finding distinct groupings or sub populations within a dataset.

```python
plt.scatter(df["mpg"], df["horsepower"], s = 12, c = "#39FF14", marker = "^")
plt.xlabel("MPG")
plt.ylabel("Horsepower")
plt.title("MPG vs Horsepower")
plt.show()
```

```python
# create figure grid with 2 column and 1 row.
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# 1. Set the global background style to dark
plt.style.use('dark_background')

axes[0].scatter(df["mpg"],df["horsepower"], c="green", marker = "^")
axes[0].set_xlabel("MPG")
axes[0].set_ylabel("Horsepower")
axes[0].set_title("MPG vs Horsepower", fontweight = "bold", fontsize = 12)
axes[1].scatter(df["displacement"],df["mpg"], c="blue", marker = "*")
axes[1].set_xlabel("Displacement")
axes[1].set_ylabel("MPG")
axes[1].set_title("Displacement vs MPG",fontweight = "bold", fontsize = 12)
plt.show()
```
